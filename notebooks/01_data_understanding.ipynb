
## Data Quality Assessment — 2009–2010

### 1. Dataset Overview

- **Dataset:** UCI Online Retail II
- **Period:** 2009–2010
- **Total Records:** 525,461
- **Total Features:** 8
- **Data Granularity:** Transaction line-item level

### 2. Data Quality Summary

| Data Quality Issue | Count | Percentage |
|:---|---:|---:|
| Missing Customer ID | 107,927 | 20.54% |
| Missing Description | 2,928 | 0.56% |
| Exact Duplicate Rows | 6,865 | 1.31% |
| Cancellation Records (Invoice starts with C) | 10,206 | 1.94% |
| Negative Quantity (Non-cancellation) | 2,121 | 0.40% |

*Note: Categories may overlap. Percentages are calculated at the transaction line-item level.*

### 3. Key Findings

**Missing Values**
- Customer ID is missing in 20.54% of records.
- Missing Customer IDs prevent reliable customer-level attribution.

**Cancellation & Inventory Adjustments**
- Cancellation records are identified by Invoice numbers starting with `C`.
- 2,121 negative-quantity records do not have a cancellation prefix.
- All 2,121 records have zero Price and missing Customer ID, suggesting possible inventory adjustments.
- One cancellation record has a positive Quantity of 1.

**Duplicate Records**
- 6,865 rows are flagged as duplicates.
- Exact duplicates may represent recording errors or legitimate repeated line items.
- Removing duplicates could affect revenue-related features.

### 4. Implications for Modeling

- Customer-level features require identifiable Customer IDs.
- Cancellation and adjustment records should not be treated as successful purchases.
- Duplicate handling requires further validation.

### 5. Proposed Next Steps

1. Assess data quality for 2010–2011.
2. Compare data quality across both years.
3. Define cleaning rules before feature engineering.

```python

from sqlalchemy import text

batch_size = 5000
total_rows = len(df_import)

# Check existing records in MySQL
with engine.connect() as conn:
    existing_rows = conn.execute(
        text("SELECT COUNT(*) FROM transactions_raw")
    ).scalar_one()

print("Existing rows:", existing_rows)
print("Expected total:", total_rows)

if not 10 <= existing_rows <= total_rows:
    raise ValueError("Unexpected table row count. Stop and inspect!")

# Import remaining records in batches
for start in range(existing_rows, total_rows, batch_size):
    batch = df_import.iloc[start:start + batch_size]

    with engine.begin() as conn:
        batch.to_sql(
            name="transactions_raw",
            con=conn,
            if_exists="append",
            index=False,
            chunksize=1000
        )

    # Show progress every 20 batches
    batch_number = (start - existing_rows) // batch_size + 1
    if batch_number % 20 == 0 or start + batch_size >= total_rows:
        print(f"Imported through row {min(start + batch_size, total_rows):,}")

print("Import loop finished!")
```
